```python
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import io

zip_path = '../Data_Folder/Large_Protein_Localization_Images.zip'

with zipfile.ZipFile(zip_path) as z:
    # Read each channel image directly out of the zip into memory
    with z.open('protein_loc_project_data/0001_blue.png') as f:
        blue = mpimg.imread(io.BytesIO(f.read()), format='png')
    with z.open('protein_loc_project_data/0001_red.png') as f:
        red = mpimg.imread(io.BytesIO(f.read()), format='png')
    with z.open('protein_loc_project_data/0001_yellow.png') as f:
        yellow = mpimg.imread(io.BytesIO(f.read()), format='png')
    with z.open('protein_loc_project_data/0001_green.png') as f:
        green = mpimg.imread(io.BytesIO(f.read()), format='png')

# Display the images
fig1 = plt.figure(figsize=(16,4))
axs = fig1.subplots(1,4)

axs[0].imshow(blue, cmap='gray')
axs[0].set_title('blue - Nucleus')
axs[0].axis('off')

axs[1].imshow(red, cmap='gray')
axs[1].set_title('red - Microtubules')
axs[1].axis('off')

axs[2].imshow(yellow, cmap='gray')
axs[2].set_title('yellow - Endoplasmic Reticulum')
axs[2].axis('off')

axs[3].imshow(green, cmap='gray')
axs[3].set_title('green - protein of interest')
axs[3].axis('off')

plt.show()
```
